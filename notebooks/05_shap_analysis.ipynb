# 05. Model Explainability via SHAP

This notebook uses **TreeSHAP (SHapley Additive exPlanations)** to explain feature contributions toward solar PV power predictions.

```python
import sys
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import shap
import joblib

# Add project root to sys.path
sys.path.append(str(Path("..").resolve()))

from src.config import CLEANED_DATA_PATH, MODELS_DIR
from src.features import create_features, get_feature_column_names
from src.explain import compute_shap_values
```

## 1. Compute SHAP Values on Test Set

```python
model = joblib.load(MODELS_DIR / "xgboost.pkl")
df_clean = pd.read_csv(CLEANED_DATA_PATH)
df_feat = create_features(df_clean)
feature_cols = get_feature_column_names()

# Daytime sample for explanation
X_daytime = df_feat[df_feat['irradiance'] > 10.0][feature_cols]
explainer, shap_values = compute_shap_values(model, X_daytime)
```

## 2. SHAP Feature Importance Summary Plot

```python
plt.figure(figsize=(10, 6))
shap.summary_plot(shap_values, X_daytime, plot_type="bar", show=False)
plt.title("SHAP Feature Importance (Bar Plot)")
plt.tight_layout()
plt.show()
```
