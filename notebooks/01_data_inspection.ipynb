# 01. NISE Dataset Inspection

This notebook inspects the real National Institute of Solar Energy (NISE) dataset `Inverter Report.xlsx`.

### Structure Discovered:
- **Sheet Name**: `Inverter Report`
- **Header Row 3**: Plant / System Name (`SURYA BHAVAN 150KW`, `NEW GUEST HOUSE`, `ADITYA BHAVAN`, `IBC`, `HIT`, `CIGS`, `CdTe`, Batteries, etc.)
- **Header Row 4**: Metric Name (`DATE`, `TIME`, `RADIATION 500 KW`, DC Voltage, DC Current, AC Power, Total Energy, Inverter Status)
- **Sampling Frequency**: 10-minute intervals

```python
import pandas as pd
import numpy as np
from pathlib import Path

file_path = Path("../data/raw/nise/Inverter Report.xlsx")

# Read raw excel header structure
df_raw = pd.read_excel(file_path, header=None)

print("Raw dataset shape:", df_raw.shape)
print("Report Title:", df_raw.iloc[0, 2])
print("Plants / Subsystems:")
plants = df_raw.iloc[3].dropna().unique()
for p in plants:
    print(" -", p)
```

```python
# Read data with header at row 4 (0-indexed index 4)
df = pd.read_excel(file_path, header=4)

print("Cleaned header shape:", df.shape)
print("Total columns:", len(df.columns))
print()
print("First 10 columns:")
print(df.columns[:10].tolist())
```

```python
# Timestamp processing & inspection
df['timestamp'] = pd.to_datetime(df['DATE'].astype(str) + ' ' + df['TIME'].astype(str), errors='coerce')
df = df.sort_values('timestamp').reset_index(drop=True)

print("Date Range:")
print("Start:", df['timestamp'].min())
print("End:", df['timestamp'].max())
print("Total Records:", len(df))
print()
print("First 5 rows (Timestamp, Radiation, Active Power sample):")
sample_cols = [c for c in ['timestamp', 'DATE', 'TIME', 'RADIATION 500 KW', 'SMA-1 ACTIVE POWER', 'DELTA-1 ACTIVE POWER '] if c in df.columns]
print(df[sample_cols].head())
```
